### Image créée par programmation

Une image peut aussi être créée pixel par pixel en utilisant une boucle d’instructions.

**Exemple**:

```python
# Création d’une image de dimension 50 x 50 avec tous les pixels noirs
image = np.zeros([50,50]) # (zero = noir)
# Mettre en blanc certains pixels
for i in range(50):
    for j in range(50):
        # pixel blanc si i+j multiple de 7
        if np.mod(i+j,7)==0:
            image[i,j] = 255
# affichage de l’image
fig, ax = plt.subplots()
ax.imshow(image,vmin=0,vmax=255)
```

#### Exercice 3
De la même manière, créez une image de taille 100 × 100 d’un disque noir de centre (xC = 50, yC = 50) et rayon r = 40 sur fond blanc :

<p style="text-align: center;">un pixel(i,j) est noir si et seulement si `(i−yC)2 +(j−xC)2 <r2 </p>


```python
# Création d’une image de dimension 100 x 100 avec tous les pixels blancs
# Dimensions de l’image
L = 100; H = 100
image = np.ones([H,L]) * 255

# Paramètres du disque
xC = 50
yC = 50
r = 40

for i in range(H):
    for j in range(L):
        #pixel noir s’il appartient au disque
        if (i-yC) ** 2 + (j-xC) ** 2 < r ** 2:
            image[j,i] = 0

fig,ax = plt.subplots()
ax.imshow(image,vmin=0,vmax=255)

```

## Correction d'une image en niveaux de gris.

On peut transformer une image en modifiant la valeur de chaque pixel par une même fonction d’une variable réelle $p \mapsto f(p)$.
**Exemple** : pour avoir le négatif d’une image, la fonction est $f(p) = 255 − p$. Dans la plupart des cas, les fonctions standard de Python ou Numpy peuvent s’appliquer directement à tous les éléments d’un tableau, sans qu’il soit nécessaire de les parcourir un par un.

**Remarque** : les images au format bitmap ne doivent en principe comporter que des valeurs entières entre 0 et 255. Néanmoins lors des calculs des valeurs non entières peuvent apparaître, et la fonction `afficher_image` permet quand même d’afficher les images correspondantes. Par contre, les valeurs doivent rester dans l’intervalle [0, 255]. Ci-dessous, un exemple de fonction ainsi qu’une procédure qui effectue la conversion en nombres compris entre 0 et 255.

```python
# définition de la fonction f
def f(p):
    return 255-p

# Tracé de la fonction
t = np.arange(256)
yt = f(t)

# lecture de l'image 
image = plt.imread("bastille.bmp")

# Opération sur image
image2 = f(image)     # appliquer la fonction
image2 =  np.clip(image2,0,255) # repasser à des valeurs entre 0 et 255 

# affichage des images
fig,ax = plt.subplots(1,3,figsize=(10,3))
ax[0].plot(t,yt,color='r') #  tracer la fonction en rouge)
ax[1].imshow(image,vmin=0,vmax=255)
ax[1].set_axis_off()
ax[2].imshow(image2,vmin=0,vmax=255)
ax[2].set_axis_off()
```

### Correction Gamma

#### Exercice 8

La correction dite Gamma s’applique aux valeurs des pixels normalisées (ramenées à l’intervalle $[0,1]$ en divisant par $255$), par l’intermédiaire de la fonction

$$ h:\left\lbrace \begin{array}{rcl} [0,1] & \rightarrow & [0,1]\\v & \mapsto & h(v)=v^\gamma \end{array} \right.$$

Le cas $\gamma = 1$ correspond à la fonction identité. La fonction $f$ à appliquer aux valeurs de pixels est donc

$$f (p) = 255 × h(p/255).$$

<img src="fonction_gamma.png" alt= “” width="300" height="300">


- Modifiez la fonction précédente afin d’appliquer cette transformation. Indication : l'instruction `x**gamma` applique la fonction $x^\gamma$ au tableau `x` tout entier.
-  Pouvez-vous prédire l’effet de la transformation pour $\gamma = 1/2$ et plus généralement pour $\gamma < 1$ ? (pensez à la position de la courbe par rapport à la diagonale). Testez par exemple sur l’image `papillon.bmp`.
- Pouvez-vous prédire l’effet de la transformation pour $\gamma = 2$ et plus généralement pour $\gamma>1$?
- Pour les images `bastille.bmp`, `desert.bmp`, `manoir.bmp`, `fruits.bmp`, testez différentes valeurs de $\gamma$ et trouvez celle qui (pour vous) donne visuellement la meilleure correction. 

```python
# Exercice 8
def modifier_image_gamma(image,gamma):
    new_image = image / 255
    new_image = 255 * (new_image ** gamma)
    return new_image

images = (bastille, desert, manoir, fruits, papillon)
gammas = (0.1,0.2,0.5,5,20)
fig,ax = plt.subplots(len(images),len(gammas) + 1,figsize=((1+len(gammas)*3),3*len(images)))
for i, image in enumerate(images):
    ax[i,0].imshow(image,vmin=0,vmax=255)
    ax[i,0].set_axis_off()
    ax[i,0].set_title('Original')
    for j,gamma in enumerate(gammas):
        ax[i,j].imshow(modifier_image_gamma(image,gamma),vmin=0,vmax=255)
        ax[i,j].set_axis_off()
        ax[i,j].set_title(f"gamma={gamma}")
    
```

## Combinaison d'images

### Opérations entre 2 images

On peut effectuer une opération entre 2 images par une fonction $f$ sur les niveaux de gris, c’est-à-dire les valeurs de pixels normalisées comme pour la correction Gamma (ramenées à l’intervalle $[0,1]$ en divisant par $255$) :
$$v' =f(v_1,v_2)$$
Par la suite, on utilisera les opérations entre images suivantes :

| Opération | Notation | Fonction correspondante |
| :-: | :-: | --- |
| produit | $F(p_1,p_2) = p_1 \otimes p_2$ | $f(v_1,v_2)=\min(v_1 \times v_2,1)$ |
| somme | $F(p_1,p_2) = p_1 \oplus p_2$ | $f(v_1,v_2)=\min(v_1 + v_2,1)$ |
| négatif | $N(p)$ | $f(v) = 1 - v$ |

La cellule ci-dessous effectue le produit des images `papillon.bmp` et `degrade-1.bmp`. 


```python
# lecture des images 
papillon = plt.imread("papillon.bmp")
degrade1 = plt.imread("degrade-1.bmp")
degrade2 = plt.imread("degrade-2.bmp")
canal_1 = plt.imread("canal_1.bmp")
canal_2 = plt.imread("canal_2.bmp")
canal_3 = plt.imread("canal_3.bmp")
desert = plt.imread("desert.bmp")
ciel = plt.imread("ciel.bmp")
masque = plt.imread("masque-desert.bmp")

# Normalisation des images 
papillon_n = papillon / 255.
degrade1_n = degrade1 / 255.

# Calcul de la nouvelle image (en clippant entre 0 et 1)
combined = np.clip(papillon_n * degrade1_n,0,1)
# On ramène la nouvelle image entre 0 et 255
combined = 255 * combined


# Affichage 
fig,ax = plt.subplots(1,3,figsize=(6,2))
ax[0].imshow(papillon,vmin=0,vmax=255)
ax[0].set_axis_off()
ax[1].imshow(degrade1,vmin=0,vmax=255)
ax[1].set_axis_off()
ax[2].imshow(combined,vmin=0,vmax=255)
ax[2].set_axis_off()

```

Créer les fonctions `produit`, `somme` et `negatif` puis calculer les images suivantes:
| `papillon.bmp` $\otimes$ `degrade-2.bmp`| produit des deux images. |
| --- | --- |
| `papillon.bmp` $\oplus$ `degrade-1.bmp`| somme des deux images. |
| `papillon.bmp` $\oplus$ `degrade-2.bmp`| somme des deux images. |
| `canal_1.bmp` $\oplus$ `canal_2.bmp`$\oplus$ `canal_3.bmp`| somme des trois images. |
| `desert.bmp` $\otimes$ `masque-desert.bmp`| produit des deux images. |
| `desert.bmp` $\otimes$ N(`masque-desert.bmp`)| produit des deux images. |

```python
# Exercice 9

### A COMPLETER - DEBUT ###
# produit entre IMAGE1 et IMAGE2
def produit(im1,im2):
    im1b = im1 / 255
    im2b = im2 / 255
    return 255 * (im1b * im2b)

def somme(im1,im2):
    return np.clip(im1*1.+im2*1., 0, 255)

def negation(im):
    return 255 - im


images = (produit(papillon,degrade2),
          somme(papillon, degrade1),
          somme(papillon,degrade2),
          somme(canal_1,somme(canal_2,canal_3)),
          produit(desert, masque),
          produit(desert, negation(masque)))
labels = ('pap x deg2', 'pap + deg1', 'pap + deg2', 'c1 + c2 + c3','desert x masque','desert x non(masque)')

# affichage des images
fig,ax = plt.subplots(1,len(images),figsize=(len(images)*2,2))
for i, image in enumerate(images):
    ax[i].imshow(image,vmin=0,vmax=255)
    ax[i].set_axis_off()
    ax[i].set_title(labels[i])
    
```
